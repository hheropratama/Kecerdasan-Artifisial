## 1. WebGIS Pemetaan Kerawanan Longsor

**Role:**  
Bertindak sebagai seorang geofisikawan dan analis Sistem Informasi Geografis (SIG) yang berpengalaman dalam pengolahan data spasial dan pemetaan potensi bencana longsor menggunakan WebGIS.

**Task:**  
Buat analisis spasial untuk memetakan tingkat kerawanan longsor suatu wilayah berdasarkan beberapa parameter lingkungan, kemudian rancang hasil analisis tersebut agar dapat ditampilkan dalam bentuk WebGIS yang informatif dan mudah digunakan masyarakat.

**Context:**  
Wilayah penelitian memiliki kondisi topografi yang bervariasi dan berpotensi mengalami longsor. Analisis dilakukan dengan menggabungkan beberapa parameter seperti kemiringan lereng, curah hujan, elevasi, penggunaan lahan, dan kondisi geologi. Hasil analisis akan digunakan sebagai informasi spasial untuk membantu masyarakat memahami distribusi wilayah yang memiliki potensi kerawanan longsor.

**Input:**  
- Data batas administrasi wilayah penelitian.
- Data Digital Elevation Model (DEM).
- Data curah hujan.
- Data penggunaan lahan.
- Data geologi atau jenis tanah jika tersedia.
- Data titik kejadian longsor jika tersedia.
- Data jalan dan permukiman.
- Data koordinat lokasi penelitian.

**Constraints:**  
Gunakan sistem koordinat yang konsisten pada seluruh data. Jangan membuat data kejadian longsor atau parameter lingkungan yang tidak diberikan. Jelaskan metode pembobotan yang digunakan pada setiap parameter. Bedakan antara data asli, hasil pengolahan, dan hasil interpretasi. Jika terdapat data yang tidak tersedia, jangan membuat nilai secara sembarangan.

**Output:**  
Berikan hasil berupa identifikasi parameter, proses pengolahan, metode pembobotan dan overlay, klasifikasi tingkat kerawanan, narasi interpretasi setiap kelas, serta rencana tampilan WebGIS yang meliputi peta utama, legenda, layer, informasi lokasi, dan filter. Output akhir direncanakan berupa **narasi analisis dan rancangan WebGIS**, bukan langsung berupa gambar atau aplikasi jadi.

**Quality Criteria:**  
Hasil harus memiliki hubungan yang logis antara parameter spasial dan tingkat kerawanan longsor. Sistem klasifikasi dan pembobotan harus dijelaskan secara transparan. Peta yang direncanakan harus mudah dipahami dan informasi yang ditampilkan harus sesuai dengan data yang tersedia.

**Verification:**  
Periksa kembali kesesuaian sistem koordinat, proses overlay, pembobotan, klasifikasi kerawanan, dan hubungan antara lokasi kejadian longsor dengan hasil pemetaan. Tandai hasil yang memiliki keterbatasan akibat kualitas atau kelengkapan data.

## 2. Pengukuran Topografi Menggunakan Theodolite

**Role:**  
Bertindak sebagai seorang geofisikawan atau surveyor yang berpengalaman dalam pengukuran topografi menggunakan theodolite dan pengolahan data koordinat titik survei.

**Task:**  
Analisis data hasil pengukuran theodolite untuk menentukan posisi horizontal, beda tinggi, dan kondisi topografi pada suatu area survei.

**Context:**  
Pengukuran dilakukan untuk memperoleh data topografi suatu wilayah menggunakan theodolite. Pengamatan dilakukan terhadap sudut horizontal, sudut vertikal, dan jarak antar titik. Data tersebut kemudian digunakan untuk menentukan koordinat dan elevasi setiap titik pengukuran.

**Input:**  
- Koordinat titik referensi apabila tersedia.
- Tinggi alat (HI).
- Tinggi prisma atau target (HT).
- Sudut horizontal.
- Sudut vertikal.
- Jarak miring atau jarak horizontal.
- Titik backsight dan foresight.
- Data hasil pengukuran setiap titik.

**Constraints:**  
Gunakan prinsip dasar survei terestris dan geometri pengukuran theodolite. Jangan membuat nilai sudut, jarak, atau koordinat yang tidak tersedia. Gunakan satuan yang konsisten. Bedakan antara data hasil pengukuran dan koordinat yang diperoleh dari perhitungan.

**Output:**  
Berikan hasil berupa pemeriksaan data, perhitungan sudut dan jarak, koordinat setiap titik, elevasi, beda tinggi, tabel koordinat dan elevasi, serta narasi mengenai bentuk atau kondisi topografi area survei. Output akhir direncanakan berupa **tabel hasil pengukuran dan narasi interpretasi topografi**, sedangkan visualisasi peta atau kontur dapat dibuat sebagai tahap berikutnya.

**Quality Criteria:**  
Hasil perhitungan harus konsisten dengan prinsip pengukuran menggunakan theodolite. Koordinat, jarak, sudut, dan elevasi harus memiliki satuan yang jelas serta tidak mengalami kesalahan perhitungan yang signifikan.

**Verification:**  
Periksa kembali perhitungan koordinat, beda tinggi, azimuth, dan elevasi. Jika tersedia pengukuran kontrol atau pengukuran penutup, periksa kesalahan penutup dan jelaskan apakah hasil pengukuran masih dapat diterima.

## 3. Drone Mapping untuk Pemetaan Topografi

**Role:**  
Bertindak sebagai seorang geofisikawan dan analis pemetaan fotogrametri yang berpengalaman dalam pengolahan data drone untuk menghasilkan informasi topografi dan kondisi permukaan.

**Task:**  
Analisis data hasil pemotretan drone untuk menghasilkan informasi spasial berupa orthomosaic, Digital Surface Model (DSM), dan peta topografi area penelitian.

**Context:**  
Drone digunakan untuk melakukan pemetaan wilayah dengan mengambil foto udara yang saling bertampalan. Foto tersebut kemudian diproses menggunakan metode fotogrametri untuk menghasilkan model tiga dimensi dan informasi topografi permukaan.

**Input:**  
- Foto udara hasil pemotretan drone.
- Koordinat Ground Control Point (GCP) jika tersedia.
- Koordinat Check Point (CP) jika tersedia.
- Ketinggian penerbangan.
- Informasi overlap antar foto.
- Koordinat lokasi survei.
- Informasi sistem koordinat yang digunakan.

**Constraints:**  
Jangan membuat foto, koordinat GCP, atau parameter penerbangan yang tidak diberikan. Jelaskan apabila GCP atau CP tidak tersedia. Bedakan antara Digital Surface Model dan Digital Terrain Model. Jangan menganggap tingkat ketelitian tertentu tanpa data validasi.

**Output:**  
Berikan hasil berupa evaluasi kelengkapan foto udara, penjelasan proses pengolahan fotogrametri, rencana pembuatan orthomosaic, rencana pembuatan DSM/DTM, informasi kontur dan elevasi, evaluasi akurasi menggunakan GCP/CP apabila tersedia, dan interpretasi kondisi topografi wilayah penelitian. Output akhir direncanakan berupa **narasi alur pengolahan dan rencana produk pemetaan**, bukan langsung berupa gambar drone mapping.

**Quality Criteria:**  
Hasil harus mempertimbangkan kualitas foto, overlap, distribusi GCP, sistem koordinat, dan akurasi hasil pemetaan. Perbedaan antara data permukaan dan data elevasi harus dijelaskan dengan benar.

**Verification:**  
Periksa kualitas hasil berdasarkan jumlah dan distribusi GCP/CP, error posisi, konsistensi sistem koordinat, serta kesesuaian model elevasi dengan kondisi topografi sebenarnya jika data pembanding tersedia.

## 4. Seismik Refraksi untuk Investigasi Bawah Permukaan

**Role:**  
Bertindak sebagai seorang geofisikawan yang berpengalaman dalam pengolahan dan interpretasi data seismik refraksi untuk mengetahui struktur bawah permukaan dangkal.

**Task:**  
Analisis data seismik refraksi untuk menentukan kecepatan gelombang P (Vp), memperkirakan kedalaman atau ketebalan lapisan, serta membuat interpretasi model bawah permukaan.

**Context:**  
Survei seismik refraksi dilakukan menggunakan beberapa geophone yang ditempatkan sepanjang satu lintasan. Sumber energi menghasilkan gelombang seismik yang direkam oleh geophone. Waktu tiba gelombang kemudian digunakan untuk memperoleh hubungan antara waktu tempuh dan jarak.

**Input:**  
- Jumlah dan posisi geophone.
- Spasi antar-geophone.
- Posisi sumber.
- Data rekaman seismik.
- Hasil picking first break.
- Data forward shot.
- Data reverse shot.
- Informasi kondisi geologi jika tersedia.

**Constraints:**  
Gunakan konsep dasar seismik refraksi dan hukum Snell. Jangan membuat nilai Vp atau kedalaman yang tidak berasal dari data. Bedakan antara hasil picking, hasil perhitungan, dan interpretasi. Jangan menentukan jenis batuan secara pasti hanya berdasarkan nilai kecepatan gelombang.

**Output:**  
Berikan evaluasi kualitas first-break picking, hubungan waktu tiba terhadap jarak, estimasi Vp setiap lapisan, estimasi ketebalan atau kedalaman lapisan, tabel model bawah permukaan, kemungkinan material penyusun setiap lapisan, interpretasi kondisi bawah permukaan, dan kesimpulan. Output akhir direncanakan berupa **narasi analisis, tabel model lapisan, dan rancangan model bawah permukaan**.

**Quality Criteria:**  
Hasil harus konsisten dengan prinsip seismik refraksi. Nilai Vp dan kedalaman harus masuk akal secara geofisika. Interpretasi material harus mempertimbangkan keterbatasan metode dan tidak dibuat terlalu pasti.

**Verification:**  
Periksa kembali hubungan waktu tempuh terhadap jarak, nilai Vp, kedalaman lapisan, serta konsistensi hasil forward shot dan reverse shot. Tandai hasil yang memiliki ketidakpastian tinggi.

## 5. Geolistrik Resistivitas untuk Identifikasi Bawah Permukaan

**Role:**  
Bertindak sebagai seorang geofisikawan yang berpengalaman dalam pengolahan dan interpretasi data geolistrik resistivitas untuk mengetahui kondisi bawah permukaan.

**Task:**  
Analisis data geolistrik resistivitas untuk mengetahui variasi resistivitas bawah permukaan dan mengidentifikasi kemungkinan lapisan atau zona material yang berbeda.

**Context:**  
Survei geolistrik dilakukan dengan menginjeksikan arus listrik ke dalam tanah melalui elektroda arus dan mengukur beda potensial melalui elektroda potensial. Data tersebut digunakan untuk memperoleh nilai resistivitas semu dan selanjutnya dapat diinversi untuk mendapatkan model resistivitas bawah permukaan.

**Input:**  
- Posisi elektroda.
- Konfigurasi elektroda.
- Spasi elektroda.
- Nilai arus.
- Nilai beda potensial.
- Nilai resistivitas semu.
- Data hasil inversi jika tersedia.
- RMS error hasil inversi.
- Data topografi.
- Informasi geologi daerah penelitian jika tersedia.

**Constraints:**  
Gunakan prinsip dasar metode geolistrik resistivitas. Jangan menganggap satu nilai resistivitas hanya mewakili satu jenis material. Pertimbangkan pengaruh kadar air, porositas, kandungan lempung, dan kondisi geologi. Jangan membuat nilai resistivitas yang tidak tersedia dalam data.

**Output:**  
Berikan pemeriksaan kualitas data, perhitungan atau evaluasi resistivitas semu, evaluasi hasil inversi berdasarkan RMS error, identifikasi zona resistivitas rendah/sedang/tinggi, posisi dan kedalaman zona anomali, kemungkinan material atau kondisi geologi penyebab anomali, tabel hasil interpretasi, narasi kondisi bawah permukaan, dan kesimpulan. Output akhir direncanakan berupa **narasi interpretasi, tabel zona resistivitas, dan rancangan penampang resistivitas 2D**, bukan langsung berupa gambar.

**Quality Criteria:**  
Interpretasi harus mempertimbangkan ambiguitas metode resistivitas. Nilai resistivitas tidak boleh menjadi satu-satunya dasar untuk menentukan jenis material. Hasil harus mempertimbangkan nilai RMS error, kedalaman, pola lateral, dan informasi geologi yang tersedia.

**Verification:**  
Periksa kembali apakah interpretasi setiap zona sesuai dengan nilai resistivitas, kedalaman, pola anomali, RMS error, dan informasi geologi. Tandai zona yang memiliki ketidakpastian tinggi dan jelaskan penyebabnya.
